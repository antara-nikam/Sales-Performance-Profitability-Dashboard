```python

import pandas as pd

# Load our retail sales dataset
df = pd.read_csv("Superstore.csv", encoding="latin1")

# Display basic information
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display the first five records
display(df.head())
```

```python

# 1. Check column names and data types
print("DATASET STRUCTURE")
df.info()

# 2. Check missing values in each column
print("\nMISSING VALUES")
print(df.isnull().sum())

# 3. Check exact duplicate rows
print("\nDUPLICATE ROWS")
print("Number of duplicate rows:", df.duplicated().sum())

# 4. Check basic statistics for numeric columns
print("\nNUMERIC SUMMARY")
display(df.describe())
```

```python
# Convert date columns into proper datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Verify the conversion
print("Order Date data type:", df["Order Date"].dtype)
print("Ship Date data type:", df["Ship Date"].dtype)

# Display a few dates
display(df[["Order Date", "Ship Date"]].head())
```

```python
# Check numeric columns for unusual values

print("Minimum and maximum values:")
display(df[["Sales", "Profit", "Quantity", "Discount"]].agg(["min", "max"]))

print("\nRows with zero or negative sales:")
print((df["Sales"] <= 0).sum())

print("\nRows with zero or negative quantity:")
print((df["Quantity"] <= 0).sum())

print("\nRows with discounts outside 0 to 1:")
print(((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
```

```python
# Understand the business coverage of the dataset

print("ORDER DATE RANGE")
print("First order:", df["Order Date"].min().date())
print("Last order:", df["Order Date"].max().date())

print("\nUNIQUE BUSINESS VALUES")
print("Categories:", df["Category"].nunique())
print("Sub-categories:", df["Sub-Category"].nunique())
print("Regions:", df["Region"].nunique())
print("Customer segments:", df["Segment"].nunique())
print("Unique customers:", df["Customer ID"].nunique())
print("Unique orders:", df["Order ID"].nunique())

print("\nCATEGORIES")
print(df["Category"].value_counts())

print("\nREGIONS")
print(df["Region"].value_counts())

print("\nCUSTOMER SEGMENTS")
print(df["Segment"].value_counts())
```

```python
# Step 5: Calculate key business performance indicators

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_quantity = df["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100
average_order_value = total_sales / total_orders

print("KEY BUSINESS KPIs")
print("-" * 35)

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Total Units Sold: {total_quantity:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Average Order Value: ${average_order_value:,.2f}")
```

```python

# Step 6: Sales and profitability by category

category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Order_Lines=("Order ID", "count")
    )
    .reset_index()
)

category_analysis["Profit_Margin_%"] = (
    category_analysis["Total_Profit"]
    / category_analysis["Total_Sales"]
) * 100

# Sort categories by sales
category_analysis = category_analysis.sort_values(
    "Total_Sales", ascending=False
)

display(
    category_analysis.round({
        "Total_Sales": 2,
        "Total_Profit": 2,
        "Profit_Margin_%": 2
    })
)
```

```python

# Step 7: Analyse profitability by sub-category

subcategory_analysis = (
    df.groupby(["Category", "Sub-Category"])
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum")
    )
    .reset_index()
)

subcategory_analysis["Profit_Margin_%"] = (
    subcategory_analysis["Total_Profit"]
    / subcategory_analysis["Total_Sales"]
) * 100

print("TOP 5 SUB-CATEGORIES BY PROFIT")
display(
    subcategory_analysis.nlargest(5, "Total_Profit")
    .round(2)
    .reset_index(drop=True)
)

print("\nBOTTOM 5 SUB-CATEGORIES BY PROFIT")
display(
    subcategory_analysis.nsmallest(5, "Total_Profit")
    .round(2)
    .reset_index(drop=True)
)
```

```python

# Step 8: Investigate discount and profitability

# Overall profitability at each discount level
discount_analysis = (
    df.groupby("Discount")
    .agg(
        Transaction_Lines=("Order ID", "count"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

discount_analysis["Profit_Margin_%"] = (
    discount_analysis["Total_Profit"]
    / discount_analysis["Total_Sales"]
) * 100

print("PROFITABILITY BY DISCOUNT LEVEL")
display(discount_analysis.round(2))

# Examine the two loss-making Furniture sub-categories
furniture_focus = df[
    df["Sub-Category"].isin(["Tables", "Bookcases"])
]

furniture_discount_analysis = (
    furniture_focus.groupby(["Sub-Category", "Discount"])
    .agg(
        Transaction_Lines=("Order ID", "count"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

furniture_discount_analysis["Profit_Margin_%"] = (
    furniture_discount_analysis["Total_Profit"]
    / furniture_discount_analysis["Total_Sales"]
) * 100

print("\nTABLES AND BOOKCASES BY DISCOUNT LEVEL")
display(furniture_discount_analysis.round(2))
```

```python

import matplotlib.pyplot as plt

# Sort discount levels from lowest to highest
chart_data = discount_analysis.sort_values("Discount")

plt.figure(figsize=(10, 5))

plt.plot(
    chart_data["Discount"] * 100,
    chart_data["Profit_Margin_%"],
    marker="o"
)

# Reference line: zero profit margin
plt.axhline(y=0, linestyle="--")

plt.title("Profit Margin by Discount Level")
plt.xlabel("Discount (%)")
plt.ylabel("Profit Margin (%)")
plt.xticks(chart_data["Discount"] * 100)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Step 10: Yearly sales and profit analysis

df["Year"] = df["Order Date"].dt.year

yearly_analysis = (
    df.groupby("Year")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Unique_Orders=("Order ID", "nunique")
    )
    .reset_index()
)

yearly_analysis["Profit_Margin_%"] = (
    yearly_analysis["Total_Profit"]
    / yearly_analysis["Total_Sales"]
) * 100

display(yearly_analysis.round(2))

# Plot yearly sales and profit
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(
    yearly_analysis["Year"].astype(str),
    yearly_analysis["Total_Sales"]
)
axes[0].set_title("Yearly Sales")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("Sales ($)")

axes[1].bar(
    yearly_analysis["Year"].astype(str),
    yearly_analysis["Total_Profit"]
)
axes[1].set_title("Yearly Profit")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Profit ($)")

plt.tight_layout()
plt.show()
```

```python

# Step 11: Monthly sales and profit analysis

df["YearMonth"] = df["Order Date"].dt.to_period("M")

monthly_analysis = (
    df.groupby("YearMonth")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

# Convert YearMonth to datetime for plotting
monthly_analysis["YearMonth"] = (
    monthly_analysis["YearMonth"].dt.to_timestamp()
)

# Plot monthly sales
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_analysis["YearMonth"],
    monthly_analysis["Total_Sales"]
)
plt.title("Monthly Sales Trend (2014–2017)")
plt.xlabel("Month")
plt.ylabel("Sales ($)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Plot monthly profit
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_analysis["YearMonth"],
    monthly_analysis["Total_Profit"]
)
plt.title("Monthly Profit Trend (2014–2017)")
plt.xlabel("Month")
plt.ylabel("Profit ($)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Highest-sales months:")
display(
    monthly_analysis.nlargest(5, "Total_Sales").round(2)
)

print("Lowest-profit months:")
display(
    monthly_analysis.nsmallest(5, "Total_Profit").round(2)
)
```

```python

# Step 12: Compare calendar months across all years

monthly_analysis["Month"] = monthly_analysis["YearMonth"].dt.month

seasonal_analysis = (
    monthly_analysis.groupby("Month")
    .agg(
        Average_Monthly_Sales=("Total_Sales", "mean"),
        Average_Monthly_Profit=("Total_Profit", "mean")
    )
    .reset_index()
)

month_names = {
    1: "January", 2: "February", 3: "March",
    4: "April", 5: "May", 6: "June",
    7: "July", 8: "August", 9: "September",
    10: "October", 11: "November", 12: "December"
}

seasonal_analysis["Month_Name"] = (
    seasonal_analysis["Month"].map(month_names)
)

seasonal_analysis = seasonal_analysis[
    ["Month", "Month_Name",
     "Average_Monthly_Sales", "Average_Monthly_Profit"]
]

print("AVERAGE PERFORMANCE BY CALENDAR MONTH")
display(seasonal_analysis.round(2))

# Plot average monthly sales
plt.figure(figsize=(12, 5))
plt.bar(
    seasonal_analysis["Month_Name"],
    seasonal_analysis["Average_Monthly_Sales"]
)
plt.title("Average Sales by Calendar Month (2014–2017)")
plt.xlabel("Month")
plt.ylabel("Average Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Step 13: Regional sales and profitability

regional_analysis = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Unique_Orders=("Order ID", "nunique")
    )
    .reset_index()
)

regional_analysis["Profit_Margin_%"] = (
    regional_analysis["Total_Profit"]
    / regional_analysis["Total_Sales"]
) * 100

regional_analysis = regional_analysis.sort_values(
    "Total_Sales", ascending=False
)

print("REGIONAL PERFORMANCE")
display(regional_analysis.round(2))

# Compare regional sales and profit
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(
    regional_analysis["Region"],
    regional_analysis["Total_Sales"]
)
axes[0].set_title("Total Sales by Region")
axes[0].set_ylabel("Sales ($)")

axes[1].bar(
    regional_analysis["Region"],
    regional_analysis["Total_Profit"]
)
axes[1].set_title("Total Profit by Region")
axes[1].set_ylabel("Profit ($)")

plt.tight_layout()
plt.show()
```

```python

# Step 14: Customer segment performance

segment_analysis = (
    df.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Unique_Orders=("Order ID", "nunique"),
        Unique_Customers=("Customer ID", "nunique")
    )
    .reset_index()
)

segment_analysis["Profit_Margin_%"] = (
    segment_analysis["Total_Profit"]
    / segment_analysis["Total_Sales"]
) * 100

segment_analysis = segment_analysis.sort_values(
    "Total_Sales", ascending=False
)

print("CUSTOMER SEGMENT PERFORMANCE")
display(segment_analysis.round(2))

# Compare sales and profit by segment
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(
    segment_analysis["Segment"],
    segment_analysis["Total_Sales"]
)
axes[0].set_title("Total Sales by Customer Segment")
axes[0].set_ylabel("Sales ($)")
axes[0].tick_params(axis="x", rotation=15)

axes[1].bar(
    segment_analysis["Segment"],
    segment_analysis["Total_Profit"]
)
axes[1].set_title("Total Profit by Customer Segment")
axes[1].set_ylabel("Profit ($)")
axes[1].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()
```

```python

# Step 15: Summarise key findings from our analysis

print("SALES PERFORMANCE & PROFITABILITY — KEY FINDINGS")
print("=" * 55)

print(
    f"1. Overall: Sales ${total_sales:,.2f}; "
    f"profit ${total_profit:,.2f}; "
    f"margin {profit_margin:.2f}%."
)

best_sales_category = category_analysis.loc[
    category_analysis["Total_Sales"].idxmax()
]
best_profit_category = category_analysis.loc[
    category_analysis["Total_Profit"].idxmax()
]
lowest_margin_category = category_analysis.loc[
    category_analysis["Profit_Margin_%"].idxmin()
]

print(
    f"2. Highest-sales category: {best_sales_category['Category']} "
    f"(${best_sales_category['Total_Sales']:,.2f})."
)
print(
    f"3. Highest-profit category: {best_profit_category['Category']} "
    f"(${best_profit_category['Total_Profit']:,.2f})."
)
print(
    f"4. Lowest-margin category: {lowest_margin_category['Category']} "
    f"({lowest_margin_category['Profit_Margin_%']:.2f}%)."
)

worst_subcategory = subcategory_analysis.loc[
    subcategory_analysis["Total_Profit"].idxmin()
]
best_region = regional_analysis.loc[
    regional_analysis["Total_Profit"].idxmax()
]
best_segment_margin = segment_analysis.loc[
    segment_analysis["Profit_Margin_%"].idxmax()
]

print(
    f"5. Largest sub-category loss: {worst_subcategory['Sub-Category']} "
    f"(${worst_subcategory['Total_Profit']:,.2f})."
)
print(
    f"6. Highest-profit region: {best_region['Region']} "
    f"(${best_region['Total_Profit']:,.2f})."
)
print(
    f"7. Highest-margin segment: {best_segment_margin['Segment']} "
    f"({best_segment_margin['Profit_Margin_%']:.2f}%)."
)
```

```python

df_mysql.columns = df_mysql.columns.str.lower()

print(df_mysql.columns.tolist())
```

```python

# Convert date columns into proper datetime values
df_mysql["order_date"] = pd.to_datetime(df_mysql["order_date"]).dt.date
df_mysql["ship_date"] = pd.to_datetime(df_mysql["ship_date"]).dt.date

print(df_mysql[["order_date", "ship_date"]].head())
print("\nDate conversion completed!")
```

```python

import numpy as np

# Convert NumPy/Pandas values into values MySQL can accept
data = [
    tuple(
        None if pd.isna(value) else value
        for value in row
    )
    for row in df_mysql.itertuples(index=False, name=None)
]

columns = ", ".join(f"`{col}`" for col in df_mysql.columns)
placeholders = ", ".join(["%s"] * len(df_mysql.columns))

sql = f"INSERT INTO superstore ({columns}) VALUES ({placeholders})"

try:
    cursor = conn.cursor()
    
    for start in range(0, len(data), 500):
        batch = data[start:start + 500]
        cursor.executemany(sql, batch)
        conn.commit()
        print(f"Inserted {min(start + 500, len(data))} / {len(data)} rows")
    
    print("All rows inserted successfully!")

except Exception as e:
    conn.rollback()
    print("Insertion error:", e)

finally:
    cursor.close()
```

```python

import pandas as pd

query = """
SELECT
    category,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS units_sold,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY category
ORDER BY total_sales DESC;
"""

category_analysis = pd.read_sql(query, conn)
category_analysis
```

```python

query = """
SELECT
    sub_category,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY sub_category
ORDER BY total_profit ASC;
"""

subcategory_analysis = pd.read_sql(query, conn)
subcategory_analysis
```

```python

query = """
SELECT
    CASE
        WHEN discount = 0 THEN '0% discount'
        WHEN discount <= 0.10 THEN '1-10%'
        WHEN discount <= 0.20 THEN '11-20%'
        WHEN discount <= 0.30 THEN '21-30%'
        ELSE 'Above 30%'
    END AS discount_group,
    COUNT(*) AS total_records,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit,
    ROUND(100 * SUM(profit) / NULLIF(SUM(sales), 0), 2) AS profit_margin_pct
FROM superstore
GROUP BY discount_group
ORDER BY MIN(discount);
"""

discount_analysis = pd.read_sql(query, conn)
discount_analysis
```

```python

query = """
SELECT
    region,
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit,
    ROUND(100 * SUM(profit) / NULLIF(SUM(sales), 0), 2) AS profit_margin_pct
FROM superstore
GROUP BY region
ORDER BY total_profit DESC;
"""

region_analysis = pd.read_sql(query, conn)
region_analysis
```

```python

query = """
SELECT
    segment,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(DISTINCT customer_id) AS total_customers,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit,
    ROUND(100 * SUM(profit) / NULLIF(SUM(sales), 0), 2) AS profit_margin_pct
FROM superstore
GROUP BY segment
ORDER BY total_profit DESC;
"""

segment_analysis = pd.read_sql(query, conn)
segment_analysis
```

```python

query = """
SELECT
    YEAR(order_date) AS order_year,
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit,
    ROUND(100 * SUM(profit) / NULLIF(SUM(sales), 0), 2) AS profit_margin_pct
FROM superstore
GROUP BY YEAR(order_date)
ORDER BY order_year;
"""

yearly_analysis = pd.read_sql(query, conn)
yearly_analysis
```

```python

query = """
SELECT
    MONTH(order_date) AS month_number,
    MONTHNAME(order_date) AS month_name,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY MONTH(order_date), MONTHNAME(order_date)
ORDER BY month_number;
"""

monthly_analysis = pd.read_sql(query, conn)
monthly_analysis
```

```python

query = """
SELECT
    product_name,
    SUM(quantity) AS units_sold,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY product_id, product_name
ORDER BY total_profit ASC
LIMIT 10;
"""

bottom_products = pd.read_sql(query, conn)
bottom_products
```

```python

query = """
WITH yearly_summary AS (
    SELECT
        YEAR(order_date) AS order_year,
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit
    FROM superstore
    GROUP BY YEAR(order_date)
)
SELECT
    order_year,
    ROUND(total_sales, 2) AS total_sales,
    ROUND(total_profit, 2) AS total_profit,
    ROUND(
        100 * total_profit / NULLIF(total_sales, 0),
        2
    ) AS profit_margin_pct,
    ROUND(
        100 * (
            total_sales - LAG(total_sales) OVER (ORDER BY order_year)
        ) / NULLIF(
            LAG(total_sales) OVER (ORDER BY order_year), 0
        ),
        2
    ) AS sales_yoy_growth_pct,
    ROUND(
        100 * (
            total_profit - LAG(total_profit) OVER (ORDER BY order_year)
        ) / NULLIF(
            LAG(total_profit) OVER (ORDER BY order_year), 0
        ),
        2
    ) AS profit_yoy_growth_pct
FROM yearly_summary
ORDER BY order_year;
"""

yearly_growth = pd.read_sql(query, conn)
yearly_growth
```
